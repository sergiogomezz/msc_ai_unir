## Ejercicio 3

Define una función dentro de la clase Numero que a partir de una cadena de caracteres, devuelve True si esa cadena de caracteres corresponde con un número romano y falso en caso contrario. Después, modifica el método para que lance un error en el caso de que el valor que nos pasan por parámetro no se corresponde con el patrón de un número romano

```python
import re

class Numero:
    def __init__(self, numero):
        self.romano = self.conversion_romano(numero)
        self.normal = numero

    def conversion_romano(self, numero):
        valores_romanos = [
            (1000, 'M'), (900, 'CM'), (500, 'D'), (400, 'CD'),
            (100, 'C'), (90, 'XC'), (50, 'L'), (40, 'XL'),
            (10, 'X'), (9, 'IX'), (5, 'V'), (4, 'IV'), (1, 'I')
        ]
        romanos = ""
        for valor, simbolo in valores_romanos:
            while numero >= valor:
                romanos += simbolo
                numero -= valor
        return romanos
    
    def imprime(self):
        print(f"{self.normal} {self.romano}")


    def suma_romano(self, romano):
        numero_sumar = self.conversion_entero(romano)
        self.normal += numero_sumar
        self.romano = self.conversion_romano(self.normal)

    def conversion_entero(self, romano):
        valores_romanos = {
            'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100,
            'D': 500, 'M': 1000
        }
        numero = 0
        prev_value = 0
        for simbolo in reversed(romano):
            valor = valores_romanos[simbolo]
            if valor <prev_value:
                numero-=valor
            else:
                numero+=valor
            prev_value=valor
        return numero

    def is_romano(self, cadena):
        if not isinstance(cadena, str):
            return False
        patron = '^M{0,3}(CM|CD|D?C{0,3})(XC|XL|L?X{0,3})(IX|IV|V?I{0,3})$'
        if re.match(patron, cadena):
            return True
        else:
            return False
            raise ValueError(f"{cadena} no es un número romano válido.")
            
```
