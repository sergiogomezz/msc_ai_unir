<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 10.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio), extrae y muestra los 100 lemas más repetidos en los comentarios de cada grupo</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta

lemas_odio = defaultdict(int)
lemas_no_odio = defaultdict(int)

for registro, intensidad in zip(doc, value):
    lemas = [token.lemma_ for token in registro if token.is_alpha] 
    if intensidad > 0:  
        for lema in lemas:
            lemas_odio[lema]+=1
    elif intensidad == 0:  
        for lema in lemas:
            lemas_no_odio[lema]+=1

rep_lemas_odio = Counter(lemas_odio).most_common(100)
rep_lemas_no_odio = Counter(lemas_no_odio).most_common(100)

print("100 lemas más repetiods en los comentarios de odio:")
for lema, cantidad in rep_lemas_odio:
    print(f"{lema}: {cantidad}")

print('----------')

print("100 lemas más repetidos en los comentarios de no odio:")
for lema, cantidad in rep_lemas_no_odio:
    print(f"{lema}: {cantidad}")
```
