```python
from scipy.stats import chi2_contingency
tabla_contingencia= [[20, 30, 15], [20, 15, 30]]
stat, p, dof, expected = chi2_contingency(tabla_contingencia)
alfa= 0.05
```

```python
p
```

```python
print('Significancia=%.3f, p=%.3f' % (alfa, p))
```

```python
if p <= alfa:
  print('Variables son asociadas (Rechazo de H0: Caracteristicas no son independientes)')
else:
  print('Variables no son asociadas (Falló Rechazo de H0: Características son independientes)')
```
